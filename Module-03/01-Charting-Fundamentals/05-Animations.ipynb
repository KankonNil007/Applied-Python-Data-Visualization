```python
# Import animation components
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.animation as animation
```

```python
# Initialize random dataset to animate
n = 100
x = np.random.randn(n)
```

```python
# Animation frame update loop callback function:
# Clears the axis, re-plots the subset of data up to the current frame index,
# and formats the title to show sample progress.
def update(curr):
    if curr == n: 
        a.event_source.stop()
    
    plt.cla()
    
    bins = np.arange(-4, 4, 0.5)

    plt.hist(x[:curr], bins=bins)

    plt.axis([-4,4,0,30])
    
    plt.gca().set_title('Sampling the Normal Distribution')
    plt.gca().set_ylabel('Frequency')
    plt.gca().set_xlabel('Value')
    plt.annotate('n = {}'.format(curr), [3,27])
```

```python
# Instantiating the animation engine:
# Calls FuncAnimation to compile the update loop and display the live simulation.
from IPython.display import HTML

a = animation.FuncAnimation(
    plt.figure(),
    update,
    frames=np.arange(0, n + 1),
    interval=100
)

HTML(a.to_jshtml())
```
