```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from datetime import datetime
```

```python
df = pd.read_csv("assets/nyc-hourly-traffic.csv")

df["Date"] = df["Date"].apply(pd.to_datetime)

df.head()
```

```python
plt.hist(df["Hour"],bins=24,weights=df["# Vehicles - E-ZPass"]);
```

```python
df["Day of Week"]=df["Date"].dt.dayofweek

# Once we have done that we can just look at a histogram
plt.hist(df["Day of Week"],bins=7,weights=df["# Vehicles - E-ZPass"]);
```

```python
plt.figure(figsize=(12,8)) # make a slightly bigger figure
plt.hist2d(df["Hour"],
           df["Day of Week"],
           bins=[24,7],
           weights=df["# Vehicles - E-ZPass"])

# This next part is optional, but adds a legend telling you the value
# of each bin (cell) in the histogram
plt.colorbar();
```
