```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
np.random.seed(1234)

v1 = pd.Series(np.random.normal(0,10,1000), name="v1")
v2 = 2*v1 + pd.Series(np.random.normal(60,15,1000), name="v2")
```

```python
plt.figure()
plt.hist(v1, alpha=0.7, bins=np.arange(-50,150,5), label='v1');
plt.hist(v2, alpha=0.7, bins=np.arange(-50,150,5), label='v2');
plt.legend();
```

```python
plt.figure()
plt.hist([v1, v2], histtype="barstacked", density=True)
v3 = np.concatenate((v1, v2))
sns.kdeplot(v3)
```

```python
plt.figure()
sns.histplot(v3, color="Teal", stat="density")
sns.kdeplot(v3, color="Navy")
plt.show()
```

```python
sns.jointplot(x=v1, y=v2, alpha=0.4);
```

```python
grid = sns.jointplot(x=v1, y=v2, alpha=0.4)
grid.ax_joint.set_aspect("equal")
```

```python
sns.jointplot(x=v1, y=v2, kind="hex");
```

```python
sns.set_style("white")

sns.jointplot(x=v1, y=v2, kind="kde", space=0)
```
