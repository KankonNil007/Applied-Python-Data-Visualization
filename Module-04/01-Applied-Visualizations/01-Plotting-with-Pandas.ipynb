```python
# Setup plotting modules and loading libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Retrieve all available visualization style sheets in Matplotlib
plt.style.available
```

```python
# Load standard color-blind style layout config
# This updates default colors, borders, and typography configs.
plt.style.use("seaborn-v0_8-colorblind")
```

```python
# Built-in Pandas Plotting API:
# Generates X/Y scatter plots directly from DataFrame columns using df.plot.scatter().
np.random.seed(123)

df = pd.DataFrame({
    "A": np.random.randn(365).cumsum(0),
    "B": np.random.randn(365).cumsum(0) + 20,
    "C": np.random.randn(365).cumsum(0) - 20
    },
    index= pd.date_range("1-1-2017", periods=365)
)

df.head()
```

```python
# Matrix subplots from Pandas DataFrames:
# Automatically layout and plot individual column variables as a grid of line charts.
df.plot()
```

```python
df.plot("A", "B", kind="scatter")
```

```python
df.plot.scatter("A", "C", c="B", s=df["B"], colormap="viridis")
```

```python
ax = df.plot.scatter("A", "C", c="B", s=df["B"], colormap="viridis")
ax.set_aspect("equal")
```

```python
df.plot.box()
```

```python
df.plot.hist(alpha=0.7)
```

```python
df.plot.kde()
```
